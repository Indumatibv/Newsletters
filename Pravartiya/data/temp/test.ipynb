```python
import re

text = "provisions88[]within six months from the date of such trigger, and the disclosures of such compliance may be made in the corporate governance compliance report on and from the third quarter89[,]"

# Old
old = re.findall(r'\b(\d+)(?=\[)', text)
print("Old:", old)  # []  -- misses 88 and 89

# New  
new = re.findall(r'(?<!\d)(\d{1,3})(?=\[)', text)
print("New:", new)  # ['88', '89']  -- correct
```
